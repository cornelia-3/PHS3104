# 4.1. Calculs analytiques

**a) Phase et interférence**

Initialement, on a $|\psi_0\rangle = |0\rangle$. Lorsqu'on applique la porte de Hadamard, on obtient

$$
|\psi_1\rangle = H|\psi_0\rangle = H|0\rangle
= \frac{1}{\sqrt{2}}\begin{pmatrix} 1 & 1 \\ 1 & -1 \end{pmatrix}\begin{pmatrix} 1 \\ 0 \end{pmatrix}
= \frac{1}{\sqrt{2}}\begin{pmatrix} 1 \\ 1 \end{pmatrix}
= \frac{1}{\sqrt{2}}\left(|0\rangle + |1\rangle\right).
$$

Ensuite, lorsqu'on applique la porte de phase $P(\varphi)=\begin{pmatrix} 1 & 0 \\ 0 & e^{i\varphi} \end{pmatrix}$ à $|\psi_1\rangle$, on obtient

$$
|\psi_2\rangle = P(\varphi)|\psi_1\rangle
= \frac{1}{\sqrt{2}}\begin{pmatrix} 1 & 0 \\ 0 & e^{i\varphi} \end{pmatrix}\begin{pmatrix} 1 \\ 1 \end{pmatrix}
= \frac{1}{\sqrt{2}}\begin{pmatrix} 1 \\ e^{i\varphi} \end{pmatrix}
= \frac{1}{\sqrt{2}}\left(|0\rangle + e^{i\varphi}|1\rangle\right).
$$

D'un côté, la probabilité de mesurer $|0\rangle$ est :

$$
P(0) = \left|\langle 0|\psi_2\rangle\right|^2
= \left|\frac{1}{\sqrt{2}}\langle 0|0\rangle + \frac{1}{\sqrt{2}}e^{i\varphi}\langle 0|1\rangle\right|^2
= \left|\frac{1}{\sqrt{2}}\right|^2 = \frac{1}{2}.
$$

D'un autre côté, la probabilité de mesurer $|1\rangle$ est :

$$
P(1) = \left|\langle 1|\psi_2\rangle\right|^2
= \left|\frac{1}{\sqrt{2}}\langle 1|0\rangle + \frac{1}{\sqrt{2}}e^{i\varphi}\langle 1|1\rangle\right|^2
= \left|\frac{1}{\sqrt{2}}e^{i\varphi}\right|^2 = \frac{1}{2}.
$$

Ensuite, lorsqu'on applique la deuxième porte de Hadamard à $|\psi_2\rangle$ pour obtenir $HP(\varphi)H|0\rangle$, on obtient

$$
|\psi_3\rangle = H|\psi_2\rangle
= \frac{1}{\sqrt{2}}\begin{pmatrix} 1 & 1 \\ 1 & -1 \end{pmatrix}\frac{1}{\sqrt{2}}\begin{pmatrix} 1 \\ e^{i\varphi} \end{pmatrix}
= \frac{1}{2}\begin{pmatrix} 1 + e^{i\varphi} \\ 1 - e^{i\varphi} \end{pmatrix}.
$$

À présent, on met $e^{i\varphi/2}$ en facteur pour obtenir

$$
|\psi_3\rangle = \frac{1}{2}e^{i\varphi/2}\begin{pmatrix} e^{-i\varphi/2} + e^{i\varphi/2} \\ e^{-i\varphi/2} - e^{i\varphi/2} \end{pmatrix},
$$

ce qui donne :

$$
|\psi_3\rangle = e^{i\varphi/2}\begin{pmatrix} \cos(\varphi/2) \\ -i\sin(\varphi/2) \end{pmatrix}
= e^{i\varphi/2}\left(\cos\!\left(\frac{\varphi}{2}\right)|0\rangle - i\sin\!\left(\frac{\varphi}{2}\right)|1\rangle\right).
$$

Ici, la probabilité de mesurer $|0\rangle$ est :

$$
P(0) = \left|\langle 0|\psi_3\rangle\right|^2
= \left|e^{i\varphi/2}\cos\!\left(\frac{\varphi}{2}\right)\langle 0|0\rangle - i\,e^{i\varphi/2}\sin\!\left(\frac{\varphi}{2}\right)\langle 0|1\rangle\right|^2
= \left|e^{i\varphi/2}\cos\!\left(\frac{\varphi}{2}\right)\right|^2
= \cos^2\!\left(\frac{\varphi}{2}\right).
$$

Ainsi, la probabilité de mesurer $|1\rangle$ est :

$$
P(1) = \left|\langle 1|\psi_3\rangle\right|^2
= \left|e^{i\varphi/2}\cos\!\left(\frac{\varphi}{2}\right)\langle 1|0\rangle - i\,e^{i\varphi/2}\sin\!\left(\frac{\varphi}{2}\right)\langle 1|1\rangle\right|^2
= \left|-i\,e^{i\varphi/2}\sin\!\left(\frac{\varphi}{2}\right)\right|^2
= \sin^2\!\left(\frac{\varphi}{2}\right).
$$

Alors, le rôle de la dernière porte de Hadamard est de créer une interférence entre les deux états $|0\rangle$ et $|1\rangle$ pour que la probabilité de mesurer $|0\rangle$ ou $|1\rangle$ dépende de la phase $\varphi$.

**b) Action de la diffusion**

Pour $|\psi\rangle = \sum_x a_x |x\rangle$ et avec $D = 2|s\rangle\langle s| - I_N$, on a

$$
D|\psi\rangle = \left(2|s\rangle\langle s| - I_N\right)|\psi\rangle = 2|s\rangle\langle s|\psi\rangle - |\psi\rangle.
$$

On regarde maintenant l'effet de $\langle s|\psi\rangle$ :

$$
\langle s|\psi\rangle = \frac{1}{\sqrt{N}}\sum_{y=0}^{N-1}\langle y|\sum_x a_x|x\rangle = \frac{1}{\sqrt{N}}\sum_x a_x.
$$

Cependant, on sait que la moyenne est $\mu = \dfrac{1}{N}\sum_x a_x$. Si on isole la somme sur $x$ des $a_x$, on obtient :

$$
N\mu = \sum_x a_x.
$$

Ainsi, $\langle s|\psi\rangle$ devient :

$$
\langle s|\psi\rangle = \frac{1}{\sqrt{N}}\, N\mu = \sqrt{N}\,\mu.
$$

En réécrivant $D|\psi\rangle$, on obtient :

$$
D|\psi\rangle = 2|s\rangle\,\mu\sqrt{N} - \sum_x a_x|x\rangle
= 2\mu\sqrt{N}\,\frac{1}{\sqrt{N}}\sum_x |x\rangle - \sum_x a_x|x\rangle
= \sum_x \left(2\mu - a_x\right)|x\rangle.
$$

Donc, l'effet de la diffusion est de transformer chaque coefficient $a_x$ en $2\mu - a_x$.

**c) Une itération à la main**

**1.**

Pour le triangle, voici respectivement les couleurs associées à $|0\rangle$ et $|1\rangle$ pour chaque sommet :

- A : rouge ($|0\rangle$), vert ($|1\rangle$)
- B : vert ($|0\rangle$), bleu ($|1\rangle$)
- C : bleu ($|0\rangle$), rouge ($|1\rangle$)

Avec $2^3 = 8$ états, on a donc

$$
|\psi_0\rangle = \frac{1}{\sqrt{8}}\left(|000\rangle + |001\rangle + |010\rangle + |100\rangle + |111\rangle + |110\rangle + |011\rangle + |101\rangle\right).
$$

Lorsqu'on analyse chaque état, on voit que les arêtes non valides sont celles où les couleurs des sommets sont les mêmes. Ainsi, les états valides sont $|000\rangle$ et $|111\rangle$. Ici, l'oracle va donc marquer les états valides en inversant le signe de leur coefficient. Ainsi, on obtient :

$$
|\psi_1\rangle = \frac{1}{\sqrt{8}}\left(-|000\rangle + |001\rangle + |010\rangle + |100\rangle - |111\rangle + |110\rangle + |011\rangle + |101\rangle\right).
$$

Pour $|000\rangle$, la probabilité est $\left|-\dfrac{1}{\sqrt{8}}\right|^2 = \dfrac{1}{8}$. Pour $|111\rangle$, la probabilité est $\left|-\dfrac{1}{\sqrt{8}}\right|^2 = \dfrac{1}{8}$. Pour les autres états, on a $\left|\dfrac{1}{\sqrt{8}}\right|^2 = \dfrac{1}{8}$.

**2.**

Après une diffusion, selon le résultat obtenu dans la section b), on a, pour le calcul de la moyenne :

$$
\mu = \frac{1}{8}\left(-\frac{2}{\sqrt{8}} + \frac{6}{\sqrt{8}}\right) = \frac{1}{8}\cdot\frac{4}{\sqrt{8}} = \frac{1}{2\sqrt{8}}.
$$

Alors, les coefficients des états valides deviennent :

$$
2\mu - a_x = 2\cdot\frac{1}{2\sqrt{8}} - \left(-\frac{1}{\sqrt{8}}\right) = \frac{1}{\sqrt{8}} + \frac{1}{\sqrt{8}} = \frac{2}{\sqrt{8}} = \frac{1}{\sqrt{2}}.
$$

Pour les autres états, on a :

$$
2\mu - a_x = 2\cdot\frac{1}{2\sqrt{8}} - \frac{1}{\sqrt{8}} = 0.
$$

Alors, le nouvel état après diffusion est :

$$
|\psi_2\rangle = \frac{1}{\sqrt{2}}\left(|000\rangle + |111\rangle\right),
$$

avec la probabilité totale :

$$
P_{\text{totale}} = P_{000} + P_{111} = \left|\frac{1}{\sqrt{2}}\right|^2 + \left|\frac{1}{\sqrt{2}}\right|^2 = \frac{1}{2} + \frac{1}{2} = 1.
$$

Les probabilités ne changent pas à l'étape de l'oracle, car seul le signe des amplitudes des états valides change (changement de phase). Comme on prend le module au carré de l'amplitude, la probabilité reste invariante.

# 4.2. Construction des opérateurs avec QuTiP

### a) Projecteurs locaux

1.

```python
def projecteur_local(n, k, b):
    #liste de n opérateurs « identité » agissant sur un qubit (dimension = 2)
    operateurs = [qeye(2)] * n 
    ket = basis(2, b) # |0> si b = 0 et |1> si b = 1
    bb = ket.proj() # |b><b|
    operateurs[k] = bb #on agit sur le k-ième qubit seulement
    projecteur = tensor(operateurs) # produit tensoriel sur toute la liste
    return projecteur
```

2. On choisit pour n = 3, k = 0 et b = 0. En d'autres termes, le projecteur local devrait conserver seulement les états dont le premier bit égale 0. On le fait agir sur les états suivants : $|010\rangle$ et $\frac{1}{\sqrt2}|110\rangle$.

```python
projecteur_test = projecteur_local(3, 0, 0)

# état conservé |010>
etat1 = basis([2]*3, [0,1,0])
#print(etat1)
print("action sur |010> : " + str((projecteur_test*etat1).full().ravel()))

# état non conservé 1/sqrt2*|110>
etat2 = 1/np.sqrt(2)*(basis([2]*3, [1,1,0]))
#print(etat2)
print("action sur 1/√2|110> : " + str((projecteur_test*etat2).full().ravel()))
```

Comme prévu, le premier état demeure inchangé alors que le deuxième donne le vecteur nul.

### b) Contraintes et oracle

1. Conflit

```python
def conflit(n, u, v, couleurs):
# on assume que couleurs est un dictionnaire des sommets
    C_uv = 0 # initialisation
    for i in [0, 1]: # sommation
        for j in [0, 1]: # double sommation
            if couleurs[u][i] == couleurs[v][j]: # on compare les couleurs des sommets
                C_uv += projecteur_local(n, k=u, b=i) * projecteur_local(n, k=v, b=j)
                # on ajoute si même couleur d'après la formule donnée
    return C_uv
```

2. Exemple de vérification : AB maison

3. Écrire oracle(graphe)

```python
def oracle(graphe):
    couleurs = graphe["couleurs"] 
    aretes = graphe["aretes"] 
    n=len(graphe["couleurs"])

    I_N = qeye([2]*n) #Matrice identite de taille 2^N par 2^N

    Pi_val = I_N #initialisation de la multiplication:
    for (u, v) in aretes: #multiplication sur toutes les arrêtes 
        C_uv = conflit(n,u,v,couleurs) # appeler la fonction de 4.2 B) 1
        Pi_val = Pi_val * (I_N - C_uv) 
    O_f = I_N - 2* Pi_val # porte finale pour l'oracle

    return O_f
```

4. Justifier l'unitarité de $O_f$

Pour vérifier si $O_f$ est unitaire, on doit vérifier si $O_f * O_f^{\dagger} = I$. $\Pi_{val}$ est un projecteur car il est construit à partir du produit des projecteurs $I_N - C_{uv}$. Ces projecteurs sont diagonaux et commutent entre eux. Leur produit est donc également un projecteur. On sait donc que: 

$$ \Pi_{val}^{\dagger} = \Pi_{val} $$
et que 
$$\Pi_{val}^2 = \Pi_{val}$$

Donc 

$$O_f^{\dagger}O_f = (I_N-2\Pi_{val})*(I_N-2\Pi_{val}) = (I_N-2\Pi_{val})^2$$

$$O_f^{\dagger}O_f = I_N^2 - 4*I_N*\Pi_{val}+4\Pi_{val}^2$$

On utilise la propriété de l'opérateur $\Pi_val^2 = \Pi_{val}$

$$O_f^{\dagger}O_f = I_N - 4*\Pi_{val}+4\Pi_{val}$$

Donc 

$$O_f^{\dagger}O_f = I_N$$

L'opérateur $O_f$ est bien unitaire. 




```python

#Vérifier numériquement que l'oracle est unitaire sur le TRIANGLE.

TRIANGLE = {"couleurs" : [("R", "V"), ("V", "B"), ("B", "R")], "aretes": [(0,1), (1,2), (2,0)]}

O_TRIANGLE = oracle(TRIANGLE)
O_TRIANGLE_dague = O_TRIANGLE.dag() #conjugue hermitien de l'oracle

X = O_TRIANGLE * O_TRIANGLE_dague # produit de l'oracle et de son conjugué hermitien
print(X.full())
```

On voit que X est la matrice identité, l'oracle est donc bien unitaire. Vérifions maintenant le signe reçu par chaque état de base du triangle. Les états de base sont les vecteurs propres de l'oracle (|x>) puisque $O_f$ est unitaire, ce qui implique que la matrice est diagonale. Les valeurs propores sont soit -1 si le coloriage est valide, soit 1 sinon (comme décrit dans la porte finale de l'énoncé). Pour une vérification indépendante des couleurs, on utilise la notation en bits.


Pour qu'il n'y ait pas de conflit, les couleurs des sommets soivent être toutes différentes, ce qui correspond aux bits 000 et 111.

```python
n = len(TRIANGLE["couleurs"]) # nombre de sommets

# en binaire, 111 correspond au nombre 7 et 000 au nombre 0.
# on peut donc faire une boucle de 0 à 7, ce qui correspond à 2**(nombre sommets)

for i in range(2**n):
    bits = [int(c) for c in format(i, "03b")] # parcourt string pour avoir liste des 3 bits
    ket = basis([2]*n, bits)
    action_oracle = O_TRIANGLE * ket
    signe = action_oracle.full().ravel()[i].real # on identifie le signe du i-eme element
    print("signe de l'etat " + format(i, "03b") + " : " + str(signe))
```

Comme attendu, seuls les états de base 000 et 111 sont valides et reçoivent le signe -1.

### C) Diffusion et circuit complet

1. Écrire Diffusion(n)

```python
H = 1/np.sqrt(2) * Qobj([[1, 1], [1, -1]]) # matrice de Hadamard 

def H_n(n):
    return tensor([H]*n) # produit tensoriel des matrice H

def diffusion(n):
    I_N = qeye([2]*n)
    bits = [0] * n # liste de n entiers 0 ou 1
    ket_zero_n = basis([2]*n, bits) #ket de n qubits |0^n>
    projecteur = ket_zero_n.proj() #projecteur |0^n><0^n|

    D = H_n(n)*(2*(projecteur)-I_N)*H_n(n) # formule pour la diffusion
    return D

```

2. Écrire circuit(graphe, r)

```python
def circuit(graphe, r):
    couleurs = graphe["couleurs"] 
    aretes = graphe["aretes"] 
    n=len(graphe["couleurs"])
    I_N = qeye([2]*n) #Matrice identite de taille 2^N par 2^N
    O_f = oracle(graphe) # Utilisation de la fonction de l'oracle définie en 4.2 b) 3. 
    D = diffusion(n) #utilisation de la fonction de diffusion définie en 4.2 c)1. 

    U_r = ((D*O_f)**r)*H_n(n) # définiition de U_r
    return U_r
    
```

Cela confirme donc que le résultat obtenu avec le circuit complet est le même que celui trouvé analytiquement à la question 4.1 c).

### b) Interférences et amplifications

1. Probabilité totale de succès

```python
# On définit les chaines valides pour la maison et pour le triangle en utilisant 
# les résultats du a)
# MAISON
chaines_valides = []
for x in range(2**5):
    bits = [int(c) for c in format(x, "05b")]
    couleurs = [MAISON["couleurs"][k][bits[k]] for k in range(5)]
    if all(couleurs[u] != couleurs[v] for u, v in MAISON["aretes"]):
        chaines_valides.append((format(x, "05b"), couleurs))
chaines_valides_maison = chaines_valides

# TRIANGLE
chaines_valides = []
for x in range(2**3):
    bits = [int(c) for c in format(x, "03b")]
    couleurs = [TRIANGLE["couleurs"][k][bits[k]] for k in range(3)]
    if all(couleurs[u] != couleurs[v] for u, v in TRIANGLE["aretes"]):
        chaines_valides.append((format(x, "03b"), couleurs))
chaines_valides_triangle = chaines_valides

# On définit les deux graphes à traiter:
graphes = [("Triangle", TRIANGLE, chaines_valides_triangle), ("Maison", MAISON, chaines_valides_maison)]

for nom, graphe, chaines_valides in graphes:
    # On commence par trouver l'état pour chaque r
    r = np.arange(0, 5)
    indices_valides = [int(c, 2) for c, _ in chaines_valides] # indices des états valides
    P_numerique = []

    for i in r:
        U_i = circuit(graphe, i) # Utiliser la fonction fait en 4.2 c) 2. pour le triangle et r=1
        n = len(graphe["couleurs"])
        bits = [0] * n
        ket_zero_n = basis([2]*n, bits) 
        Psi_i = U_i *  ket_zero_n 

        # On cherche maintenant la probabilité de chaque état de base pour chaque r
        amplitude_numerique = Psi_i.full().ravel() #amplitude de Psi sous forme de tableau, dans l'ordre des indices
        probabilite = np.abs(amplitude_numerique)**2 #probabilité de chaque état de base

        # On somme les probabilités sur les chaînes valides
        probabilite_somme = sum(probabilite[indice] for indice in indices_valides)
        P_numerique.append(probabilite_somme)

    # On écrit la formule théorique pour la probabilité de succès:
    theta = np.arcsin(np.sqrt(len(chaines_valides)/2**n))
    P_theorique = [np.sin((2*i+1)*theta)**2 for i in r]

    # On peut calculer le point d'arret optimal théorique et le point d'arrêt optimal numérique
    r_optimal_theorique = round((np.pi/(4*theta) - 1/2))

    r_optimal = r[-1]
    for k in range(len(P_numerique)-1):
        if P_numerique[k+1] <= P_numerique[k]:
            r_optimal = k
            break

    # On trace maintenant les points numériques et théoriques sur le même graphique
    print('Probabilité de succès pour', f"{nom}")
    for i in r:
        print(f"r = {i} : numérique = {P_numerique[i]:.4f}, théorique = {P_theorique[i]:.4f}")
    print(f"Point d'arrêt optimal numérique : {r_optimal}, théorique : {r_optimal_theorique}") 

    plt.figure(figsize=(10, 6))
    plt.plot(r, P_numerique, 'o-', label='Probabilité numérique', color='blue')
    plt.plot(r, P_theorique, 's--', label='Probabilité théorique', color='orange')
    plt.axvline(x=r_optimal, color='green', linestyle='--', label=f'Point d\'arrêt optimal (numérique): {r_optimal}')
    plt.axvline(x=r_optimal_theorique, color='red', linestyle='--', label=f'Point d\'arrêt optimal (théorique): {r_optimal_theorique}')
    plt.title(f'Graphe {nom} : Probabilité de succès en fonction du nombre d\'itérations r')
    plt.xlabel('Nombre d\'itérations r')
    plt.ylabel('Probabilité de succès')
    plt.legend()
    plt.xticks(r)
    plt.show()
```

3. Explications <br>

Premièrement, l'oracle modifie le signe des amplitudes des états en créant un décalage de phase. Les chaînes valides (v.p. -1) sont décalées par rapport à celles non valides (v.p. 1), mais la probabilité d'occurence de tous les états demeure la même (probabilité demeure uniforme, comme avant), car il s'agit de la norme au carré de l'amplitude et $|a|^2=|-a|^2$. La moyenne de l'amplitude des états ($\mu$) baisse à cause de ceux valides qui deviennent négatifs. Deuxièmement, la diffusion crée une interférence constructive entre les amplitudes des solutions valides et destructive entre celles des solutions non valides. En effet, les solutions, dont le signe est inversé se retrouvent loin de $\mu$, si bien que l'action $a_{x_{sol}}$ --> $2\mu-a_{x_{sol}}$ = $2\mu+a_x$ les gardent au-dessus de la moyenne et augmentent leur amplitude. Quant à eux, les états non solutions, dont l'amplitude est proche de $\mu$, y demeurent proches et diminuent par rapport à $\mu$ (--> $2\mu-a_x$). Toutefois, une subtilité importante de ce phénomène de diffusion est que c'est un effet périodique, si bien qu'après avoir rotationné un état à chaque itération, il se peut que les solutions s'éloignent de leur maximum et que les états non-valides s'éloignent de la probabilité nulle d'occurence. On peut voir le problème ainsi avec la formule donnée: chaque itération est une rotation d’angle $2\theta$ dans le plan solutions/non-solutions, donc la probabilité $P = \sin^2((2r+1)\theta)$ est périodique en $r$ et dépasse son maximum si on continue.
<br>
Voyons pour chacun des cas étudiés.
<br>

**Triangle.** Le triangle compte $N = 2^3 = 8$ affectations, dont $M = 2$ valides (000 et 111). Ainsi,
$$\sin\theta = \sqrt{M/N} = \tfrac{1}{2} \quad\Rightarrow\quad \theta = \tfrac{\pi}{6}.$$
Pour $r = 1$, $(2r+1)\theta = \tfrac{\pi}{2}$, donc $P_{\text{succès}}(1) = \sin^2\!\left(\tfrac{\pi}{2}\right) = 1$ : le succès est certain après une seule itération, ce qui concorde avec la valeur numérique. Pour un r plus grand, la valeur redescend!

**Maison.** La maison compte $N = 2^5 = 32$ affectations, dont $M = 4$ valides. Ainsi
$$\theta = \arcsin\sqrt{M/N} = \arcsin\sqrt{1/8} \approx 0{,}361\ \text{rad}, \qquad \frac{\pi}{4\theta} - \frac{1}{2} \approx 1{,}67,$$
donc $r^* = 2$. L'angle $(2r+1)\theta$ vaut alors $\approx 1{,}81$ rad, le plus proche de $\pi/2$, et $P_{\text{succès}}(2) = \sin^2(5\theta) \approx 0{,}945$. À $r = 3$, l'angle vaut $\approx 2{,}53$ rad, dépasse $\pi/2$, et la probabilité redescend à $\approx 0{,}33$.
