# 1.1 Calculs analytiques

a) Phase et interférence
 
Initialement, on a |psi_0>=|0>. Lorsqu'on applique la porte de Hadamard,
on obtient |psi_1> = H|psi_0> = H|0> = 1/sqrt(2)[1 1/1 -1][1 0]^T
= 1/sqrt(2)[1 1]^T = 1/sqrt(2)(|0> + |1>). Ensuite, lorsqu'on applique 
la porte de phase P(phi)=[ 1 0/ 0 exp(iphi)] à |psi_1>, on obtient
|psi_2> = P(phi)|psi_1> = 1/sqrt(2)[ 1 0/ 0 exp(iphi)][1 1]^T = 
1/sqrt(2)[1 exp(iphi)]^T = 1/sqrt(2)(|0> + exp(iphi)|1>). Ici, la
probabilité de mesurer |0> est : P(0) = |<0|psi_2>|^2 = 
|1/sqrt(2)<0|0> + 1/sqrt(2)exp(iphi)<0|1>|^2 = |1/sqrt(2)|^2 = 1/2.
D'un autre côté, la probabilité de mesurer |1> est : P(1)
= |<1|psi_2>|^2 = |1/sqrt(2)<1|0> + 1/sqrt(2)exp(iphi)<1|1>|^2 =
|1/sqrt(2)exp(iphi)|^2 = 1/2.

Ensuite, lorsqu'on applique la deuxième porte de Hadamard à |psi_2> 
pour obtenir HP(phi)H|0>, on obtient |psi_3> = H|psi_2> = 
1/sqrt(2)[1 1/1 -1]1/sqrt(2)[1 exp(iphi)]^T = 
1/2[1 + exp(iphi)/1 - exp(iphi)]^T
À présent, on met exp(iphi/2) en facteur pour obtenir :
|psi_3> = 1/2*exp(iphi/2)[exp(-iphi/2) + 
exp(iphi/2)/exp(-iphi/2) - exp(iphi/2)]^T
Ce qui donne : |psi_3> = exp(iphi/2)[cos(phi/2)/-i*sin(phi/2)]^T=
exp(iphi/2)(cos(phi/2)|0> - i*sin(phi/2)|1>)
Ici, la probabilité de mesurer |0> est : P(0) = |<0|psi_3>|^2 =
|exp(iphi/2)cos(phi/2)<0|0> - i*exp(iphi/2)sin(phi/2)<0|1>|^2 = 
|exp(iphi/2)cos(phi/2)|^2 = cos^2(phi/2).
Ainsi, la probabilité de mesurer |1> est : P(1) = |<1|psi_3>|^2 =
|exp(iphi/2)cos(phi/2)<1|0> - i*exp(iphi/2)sin(phi/2)<1|1>|^2 = 
|-iexp(iphi/2)sin(phi/2)|^2 = sin^2(phi/2).
Alors, le rôle de la dernière porte de Hadamard est de créer 
une interférence entre les deux états |0> et |1> pour que
la probabilité de mesurer |0> ou |1> dépende de la phase phi.

b) Action de la diffusion

Pour |psi>=somme sur x de a_x|x> et avec D=2|s><s| - I_N, on a 
D|psi> = (2|s><s| - I_N)|psi> = 2|s><s|psi> - |psi>. On regarde maintenant
l'effet de <s|psi> : <s|psi> = 1/sqrt(N)somme sur y=0 à N-1 de 
<y|somme sur x de a_x|x>. Cependant, on sait que la moyenne mu = 
1/N somme sur x de a_x. Si on isole la somme sur x de a_x, on obtient :
N*mu=somme sur x des a_x. Ainsi, <s|psi> devient : 1/sqrt(N)* N* mu =
sqrt(N)*mu. En réécrivant D|psi>, on obtient : D|psi> = 2|s> mu*sqrt(N)-
somme sur x de a_x|x> = 2*mu*sqrt(N)1/sqrt(N)somme sur x de |x> - 
somme sur x de a_x|x> = somme sur x de (2*mu - a_x)|x>. Donc, 
l'effet de la diffusion est de transformer chaque coefficient a_x en
2*mu - a_x.


c) Une itération à la main

 1. 
Pour le triangle, voici respectivement les sommets, |0>, |1> :
A rouge vert
B vert bleu
C bleu rouge
Avec 2^3=8 états, on a donc |psi_0> = 1/sqrt(8)(|000> + |001> + 
|010> + |100> + |111> + |110> + |011> + |101>). Lorsqu'on analyse chaque 
état, on voit que les arrêtes non valides sont ceux où les couleurs 
des sommets sont les mêmes. Ainsi, les états valides sont : |000> et |111>. 
Ici, l'oracle va donc marquer les états valides en inversant le signe 
de leur coefficient. Ainsi, on obtient : |psi_1> = 1/sqrt(8)(-|000> +
+ |001> + |010> + |100> - |111> + |110> + |011> + |101>). Pour |100>,
l'amplitude est |-1/sqrt(8)|^2 = 1/8. Pour |111>, l'amplitude est 
|-1/sqrt(8)|^2 = 1/8. Pour les autres amplitudes, on a |1/sqrt(8)|^2 = 1/8.  

 2.

Après une diffusion, selon le résultat obtenu dans la section b), on a,
pour le calcul de la moyenne : mu = 1/8(-2/sqrt(8) + 6/sqrt(8)) = 
1/8(4/sqrt(8)) = 1/(2*sqrt(8)). Alors, les coefficients des états valides 
deviennent : 2*(1/(2*sqrt(8))) - (-1/sqrt(8)) = 1/sqrt(8) + 1/sqrt(8) = 
2/sqrt(8) = 1/sqrt(2) . Pour les autres états, on a :
2*(1/(2*sqrt(8))) - (1/sqrt(8)) = 0. Alors, le nouvel état après diffusion est :
|psi_2> = 1/sqrt(2)(|000> + |111>) avec la probabilité totale:
P_totale = P_(000) + P_(111) = |1/sqrt(2)|^2 + |1/sqrt(2)|^2 = 1/2 + 1/2 = 1.
Les probabilités ne changent pas à l'étape de l'oracle, car ce n'est que 
l'amplitude des états valides qui changent de signe (changement de phase),
mais comme le module au carré de l'amplitude est prit, alors au final la 
probabilité se voit invariante.
